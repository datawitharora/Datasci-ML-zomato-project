```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Ensure 'Rating' column is numeric, coercing errors to NaN
zomato_reviews_df['Rating'] = pd.to_numeric(zomato_reviews_df['Rating'], errors='coerce')

# Drop rows where 'Rating' is NaN after conversion for this analysis
cleaned_zomato_reviews_df = zomato_reviews_df.dropna(subset=['Rating']).copy()

restaurant_review_counts = cleaned_zomato_reviews_df['Restaurant'].value_counts()
average_ratings = cleaned_zomato_reviews_df.groupby('Restaurant')['Rating'].mean().sort_values(ascending=False)

# Combine average ratings and review counts
engagement_df = pd.DataFrame({'Average_Rating': average_ratings, 'Review_Count': restaurant_review_counts})

# Fill any potential NaN values that might arise from restaurants with no reviews or ratings (though unlikely with current data)
engagement_df.fillna(0, inplace=True)

# Calculate an engagement score (e.g., product of average rating and review count)
engagement_df['Engagement_Score'] = engagement_df['Average_Rating'] * engagement_df['Review_Count']

# Sort by engagement score in descending order and get the top 10
top_10_engaged_restaurants = engagement_df.sort_values(by='Engagement_Score', ascending=False).head(10)

# Visualize the top 10 restaurants by Engagement Score
plt.figure(figsize=(14, 8))
sns.barplot(x='Restaurant', y='Engagement_Score', data=top_10_engaged_restaurants, palette='coolwarm', hue='Restaurant', legend=False)
plt.title('Top 10 Restaurants by Engagement Score')
plt.xlabel('Restaurant Name')
plt.ylabel('Engagement Score (Avg Rating * Review Count)')
plt.xticks(rotation=45, ha='right') # Rotate labels for better readability
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout() # Adjust layout to prevent labels from being cut off
plt.show()
```

## ***5. Hypothesis Testing***

Hypothesis testing is a statistical method used to determine if there is enough evidence in a sample data to infer that a certain condition is true for the entire population. It typically involves formulating two competing hypotheses:

*   **Null Hypothesis (H0):** A statement of no effect or no difference. It represents the status quo or what is generally accepted.
*   **Alternative Hypothesis (H1 or Ha):** A statement that contradicts the null hypothesis, suggesting that there is an effect or a difference.

The goal is to determine whether the evidence supports rejecting the null hypothesis in favor of the alternative hypothesis.

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

For this hypothetical statement, we will examine if the number of followers a reviewer has affects the average rating of a restaurant. We will use simple linear regression to model this relationship.

*   **Null Hypothesis (H0):** There is no significant linear relationship between the average number of followers a reviewer has and the average rating of a restaurant. (i.e., the slope coefficient for 'Avg_Reviewer_Followers' is zero, β1 = 0).
*   **Alternative Hypothesis (H1):** There is a significant positive linear relationship between the average number of followers a reviewer has and the average rating of a restaurant. (i.e., the slope coefficient for 'Avg_Reviewer_Followers' is greater than zero, β1 > 0).

First research hypothesis clearly, identifying both the null (H0) and alternative (H1) hypotheses.

* Null hypothesis: The number of followers a reviewer has has no effect on the rating of a restaurant. (H0: 𝛽1 = 0)
* Alternative hypothesis: Alternative Hypothesis: The number of followers a reviewer has has a positive effect on the rating of a restaurant. (H1: 𝛽1 > 0)
* Test : Simple Linear Regression test

#### Example:
**H0:** There is no significant difference in the average rating between restaurants in a specific cuisine category and the overall average rating.
**H1:** There is a significant difference in the average rating between restaurants in a specific cuisine category and the overall average rating.

### K-Means Clustering Model Explanation and Evaluation

K-Means is a popular unsupervised machine learning algorithm used for clustering. It aims to partition `n` observations into `k` clusters, where each observation belongs to the cluster with the nearest mean (centroid). The goal is to define the clusters such that the total within-cluster sum of squares (WSS) is minimized.

**Key Steps:**
1.  **Initialization:** Choose the number of clusters `k` and randomly initialize `k` centroids.
2.  **Assignment:** Assign each data point to the closest centroid.
3.  **Update:** Recalculate the centroids as the mean of all data points assigned to that cluster.
4.  **Iteration:** Repeat steps 2 and 3 until the centroids no longer change significantly or a maximum number of iterations is reached.

### Evaluation Metrics for K-Means

Since K-Means is an unsupervised algorithm, there's no ground truth to compare against. Evaluation often involves assessing the compactness and separation of clusters.

1.  **Elbow Method (for optimal `k`):** This method involves plotting the within-cluster sum of squares (inertia) as a function of the number of clusters (`k`). The 'elbow point' on the plot, where the rate of decrease in inertia sharply changes, is often considered the optimal `k`.
2.  **Silhouette Score:** This metric measures how similar an object is to its own cluster compared to other clusters. The silhouette score ranges from -1 to 1, where:
    *   **+1:** Indicates that the object is far away from the neighboring clusters.
    *   **0:** Indicates that the object is on or very close to the decision boundary between two neighboring clusters.
    *   **-1:** Indicates that the object might have been assigned to the wrong cluster.

#### Elbow Method to Determine Optimal K

We will use the Elbow Method to find the optimal number of clusters (`k`). We'll compute the inertia (within-cluster sum of squares) for a range of `k` values and plot them.

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Determine the optimal number of clusters using the Elbow Method
wss = [] # Within-Cluster Sum of Squares
for i in range(1, 11): # Test k from 1 to 10
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42, n_init=10)
    kmeans.fit(scaled_features_df)
    wss.append(kmeans.inertia_)

# Plot the Elbow Method graph
plt.figure(figsize=(10, 6))
plt.plot(range(1, 11), wss, marker='o', linestyle='--')
plt.title('Elbow Method for Optimal K')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Within-Cluster Sum of Squares (WSS)')
plt.grid(True)
plt.show()
```

#### Silhouette Score to Determine Optimal K

Next, we'll use the Silhouette Score, which provides a more quantitative measure of cluster quality.

```python
from sklearn.metrics import silhouette_score

# Calculate Silhouette Scores for different k
silhouette_scores = []
for i in range(2, 11): # Silhouette score requires at least 2 clusters
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42, n_init=10)
    kmeans.fit(scaled_features_df)
    score = silhouette_score(scaled_features_df, kmeans.labels_)
    silhouette_scores.append(score)

# Plot the Silhouette Scores
plt.figure(figsize=(10, 6))
plt.plot(range(2, 11), silhouette_scores, marker='o', linestyle='--')
plt.title('Silhouette Score for Optimal K')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Silhouette Score')
plt.grid(True)
plt.show()
```
